```python
import os
import json
import pandas as pd
```

```python
def get_average_recall(model,type="node"):
    sum_recall = 0
    sum_precision = 0
    sum_f1 = 0
    total_file = 0
    path = f"./uiuc_diagram_evaluater/eval_output/{model}"
    for i in os.listdir(path):
        if i.endswith(".json"):
            with open (path + "/" + i, "r") as f:
                data = json.load(f)
                scores = data[f'{type}_alignment']['scores']
                sum_recall+= scores['recall']
                sum_precision+= scores['precision']
                sum_f1 += scores['f1']
                total_file+=1
    return {"precision": sum_precision/total_file, "recall": sum_recall/total_file, "f1": sum_f1/total_file}
```

```python
node_output = {"claude":0, "gpt4o":0,"gpt5":0}
for m in list(node_output.keys()):
    node_output[m] = get_average_recall(m)
```

```python
path_output = {"claude":0,"gpt4o":0,"gpt5":0}
for m in list(path_output.keys()):
    path_output[m] = get_average_recall(m,type="path")
```

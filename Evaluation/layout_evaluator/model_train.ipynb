```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import joblib
```

```python
# Load dataset
# df = pd.read_csv('./OpenScience-UMLLayoutQuality-1.2/03-features-labels/dataset.csv')
df = pd.read_csv("./my_features_with_scores.csv")
X = df.drop(columns=['name','Quality'])
y = df['Quality']

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=41)
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.neural_network import MLPRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, ExtraTreesRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.gaussian_process import GaussianProcessRegressor

# Define models
models = {
    "RandomForest": RandomForestRegressor(n_estimators=300, random_state=41),
    "GradientBoosting": GradientBoostingRegressor(random_state=41),
    "ExtraTrees": ExtraTreesRegressor(n_estimators=300, random_state=41),
    "DecisionTree": DecisionTreeRegressor(random_state=41),
    "KNN": KNeighborsRegressor(n_neighbors=5),
    "Ridge": Ridge(alpha=1.0),
    "Lasso": Lasso(alpha=0.01),
    "ElasticNet": ElasticNet(alpha=0.01, l1_ratio=0.5),
    "GaussianProcess": GaussianProcessRegressor(),
    "SVM": SVR(kernel="rbf", C=10, epsilon=0.1),
    "NeuralNet": MLPRegressor(hidden_layer_sizes=(64,32), max_iter=1000, random_state=41)
}

results = {}

# Train and evaluate each
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    results[name] = {"MSE": mse, "RMSE": rmse, "R2": r2}

# Print results
for name, metrics in results.items():
    print(f"{name}: MSE={metrics['MSE']:.3f}, RMSE={metrics['RMSE']:.3f}, R2={metrics['R2']:.3f}")


```

```python
models = RandomForestRegressor(
    n_estimators=500,       
    max_depth=None,         
    min_samples_split=5,   
    min_samples_leaf=2,
    max_features="sqrt",
    random_state=42
)
models.fit(X_train, y_train)
y_pred = models.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
print(f"Optimized RandomForest: MSE={mse:.3f}, RMSE={rmse:.3f}, R2={r2:.3f}")
```
